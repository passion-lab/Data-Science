# Mini-Project: Sensor Data Analysis

## Scenario:
You’re analyzing data from a wearable fitness tracker. The raw data is a 1D array of heart rates (beats per minute) recorded every minute for 24 hours (1440 values).

## Tasks:
1. Reshape the data into a 24x60 matrix (rows = hours, columns = minutes).
2. Calculate the average heart rate per hour.
3. Normalize each hour’s data by subtracting its mean.

## Sample Data:
```python
heart_rates = np.random.randint(60, 100, size=1440)  # Simulated data
```

## Hints:
- Use `reshape(24, 60)`.
- Use `axis=1` for row-wise operations.

````python
import numpy as np
````

````python
# Heart rate (BMP) data from a wearable fitness tracker for 24 Hrs. recorded every minutes
heart_rates = np.random.randint(55, 111, size=1440)  # Simulated data
heart_rates_per_hour = heart_rates.reshape(24, 60)  # Making the data into hours (row) and minutes (column) format

# Calculating the average heart rate per hour (to maintain shape for broadcasting 'keepdims=True' is used)
avg_heart_rate_per_hour = np.mean(heart_rates_per_hour, axis=1, keepdims=True)
print("Average heart rate per hour are:\n", avg_heart_rate_per_hour)

# Normalizing each hour's data by subtracting its mean
normalized_heart_rates_per_hour = heart_rates_per_hour - avg_heart_rate_per_hour
print("Normalized heart rates per hour:\n", normalized_heart_rates_per_hour)
````

## Detect Anomalies:
Flag hours where the average heart rate is outside the normal range (e.g., < 65 or > 95 BPM):

````python
anomalies = np.where((avg_heart_rate_per_hour < 65) | (avg_heart_rate_per_hour > 95))[0]
print("Anomalous hours:", anomalies)
````

## Visualize Trends:
Use Matplotlib to plot hourly averages

````python
import matplotlib.pyplot as plt
````

````python
plt.plot(np.arange(24), avg_heart_rate_per_hour, marker='o')
plt.xlabel("Hour of Day (Hrs.)")
plt.ylabel("Average Heart Rate (BPM)")
plt.title("Hourly Heart Rate Trends")
plt.show()
````

## Hourly Variability:
Calculate the standard deviation of heart rates for each hour to measure variability:

````python
hourly_std = np.std(heart_rates_per_hour, axis=1)
print("Hourly variability (std) of heart rates:", hourly_std)
````
