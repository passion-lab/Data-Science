# Coin flip simulation

## Tasks:
- Simulate 1000 coin flips and calculate the probability of getting heads.
- Plot the number of heads

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Simulating 1000 coin flips using 0s and 1s
coin_flips = np.random.randint(0, 2, size=1000)

# Calculating the total number of heads
# Assuming: 0 = tails, 1 = heads
# Use 'np.sum()' if the values are only 0s and 1s to get the total number of 1s as their sum, otherwise
# ...use the below method to get the count of a particular value in an NumPy array
total_heads = coin_flips[coin_flips == 1].size

# Calculating the pronanlity of getting heads
# Formula: Probablity = Total Number of Heads / Total Number of Flips
heads_probablity = (total_heads / 1000) * 100  # Multiplied by 100 to get the percentage value

print(f"Probablity of getting heads from a 1000 flips of a coin is: {heads_probablity:.1f}%")
```

```python
# Count Heads and Tails
counts = [total_heads, 1000 - total_heads]
outcomes = [f"Heads ({counts[0]})", f"Tails ({counts[1]})"]

# Plot
plt.pie(counts, labels=outcomes, colors=["blue", "orange"])
plt.title("Coin Flip Simulation (1000 Flips)")
plt.show()
```
