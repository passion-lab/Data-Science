# NumPy (Numerical Python)

````python
import numpy as np
````

# Step 1: Creating and Manipulating Arrays

### Concept:
- A NumPy array is like a supercharged list that can handle math operations on entire datasets at once.
- Think of it as a grid of numbers (like an Excel sheet) optimized for speed.
### Real-World Connection:
Arrays of zeros or ones are used to initialize matrices in machine learning models, like neural networks!

````python
# Array of zeros
zeros = np.zeros((2, 3))  # 2 rows, 3 columns
print("Zeros:\n", zeros)

# Array of ones
ones = np.ones(12)  # One row array with 12 data
print("Ones:\n", ones)

# Array with a range
range_arr = np.arange(0, 10, 2)  # Start at 0, stop before 10, step by 2
print("Range:\n", range_arr)
````

# Step 2: Basic Operations
- NumPy makes math with arrays simple and fast.
- Unlike Python lists, NumPy applies operations to every element at once—no loops needed!

````python
arr = np.array([1, 2, 3, 4])

# Add, subtract, multiply, divide to each element of an array by a number
print("Add 5:", arr + 5)      # [6, 7, 8, 9]
print("Multiply by 2:", arr * 2)  # [2, 4, 6, 8]

# Element-wise (with same size ans shape) operations between arrays
arr2 = np.array([5, 6, 7, 8])
print("Add arrays:", arr + arr2)  # [6, 8, 10, 12]
````

````python
# Create a 1D array from a Python list
temperatures = [25, 28, 30, 22, 19]
numpy_temps = np.array(temperatures)

# Convert Celsius to Fahrenheit in one line!
fahrenheit_temps = numpy_temps * 9/5 + 32

print("Celsius:", numpy_temps)
print("Fahrenheit:", fahrenheit_temps)
````

# Step 3: Multi-Dimensional Arrays

### Concept:
- Real-world data is rarely 1D (e.g., images are 2D grids, videos are 3D).
- NumPy handles n-dimensional arrays effortlessly.

````python
# Create a 2D array (rows = cities, columns = months)
monthly_temps = np.array([
    [25, 28, 30],  # City A (Jan, Feb, Mar)
    [18, 22, 20],  # City B
    [12, 15, 14]   # City C
])

print("2D Array of temp of three cities:\n", monthly_temps)
````

````python
# Calculate average temperature for each city (use axis=1 to calcute the value in each row)
avg_monthly_temps = np.mean(monthly_temps, axis=1)
print("Average temps per city: ", avg_monthly_temps)
````

````python
# Maximum temp in each cities
np.max(monthly_temps, axis=1)
````

````python
# Find the hottest month across all cities
np.argmax(monthly_temps)  # output: 2 (index 2 referes to: City A, Mar)
````

````python
monthly_temps_2 = np.array([
    [25, 28, 30],  # City A (Jan, Feb, Mar)
    [18, 32, 20],  # City B
    [12, 15, 14]   # City C
])
# Find the hottest month across all cities
np.argmax(monthly_temps_2)  # output: 4 (index 4 referes to: City B, Feb)
````

# Real-World Mini-Project: Analyze Sports Data
## Task:
You’re a data analyst for a football team. The coach wants to know:
1. Which player has the most consistent performance (lowest standard deviation in goals per match)?
2. What’s the total goals scored in the last 5 matches?

## Date:
```python
goals = np.array([
    [2, 0, 1, 3, 2],  # Player 1
    [1, 1, 0, 2, 1],  # Player 2
    [3, 2, 2, 0, 4]   # Player 3
])
```
## Hints:
- Use `np.std()` for standard deviation.
- Use slicing `([:, -5:])` to get the last 5 matches.

````python
# Given data of Goals for three football players
goals = np.array([
    [2, 0, 1, 3, 2],  # Player 1
    [1, 1, 0, 2, 1],  # Player 2
    [3, 2, 2, 0, 4]   # Player 3
])

# Add Assertions: Validate inputs (to ensure 5 matches).
assert goals.shape[1] == 5

# Finding the most consistent player (who has the lowest value) with Standard Deviation
# IDEA: the less the value of SD, the more the player has consistant performance
player_sds = np.std(goals, axis=1)
print("Standard Deviations of goals for each palyer:", player_sds)
# Finding the minimum SD value by considering lowest_sd as a 1-D array.
most_consistent_player = np.argmin(player_sds) + 1  # +1 used to make it Non-Zero based index (1, 2, 3... instead of 0, 1, 2, ...)
print(f"Player with most consistent performances is Player {most_consistent_player} as of his least SD value.")

# Finding the total goals scored in the last 5 matches
total_goals_in_last_5_matches = np.sum(goals[:, -5:])
print("Total goals scored in the last five matches are:", total_goals_in_last_5_matches)
````

# Key Takeaways for NumPy
- **Vectorization**: Always prefer array operations over loops (faster and cleaner).
- **Axis Understanding**:
    - `axis=1` = row-wise operations.
    - `axis=0` = col-wise operations.
- **Edge Cases**: Some methods returns only the first occurance (if multiple exist) e.g., `np.argmax()`
    - *Solutions:*
        1. **Find All Indices** with `np.where()` where the value equals the maximum:
           ```python
           scores = np.array([90, 90, 85, 90, 75])
           max_val = scores.max()
           all_max_indices = np.where(scores == max_val)[0]  # Output: [0, 1, 3]
           ```
        2. **Random Selection** randomly pick from `all_max_indices`:
           ```python
           np.random.choice(all_max_indices)  # Could return 0, 1, or 3
           ```
        3. **Weighted Preference** for tie-breaking, apply custom logic (e.g., prioritize later indices):
           ```python
           last_max_index = all_max_indices[-1]  # Output: 3
           ```
    - *Real-World Application:*
    In recommendation systems, if two movies have the same rating, you might want to show both or prioritize the newer one.

## Topic 1: Reshaping & Slicing Arrays

1. Why Learn This?
    - Data often needs restructuring (e.g., converting a 1D sensor stream into a 2D time-series matrix).
2. Concept:
    - Use `.reshape()` to change array dimensions without altering data.
    - Total elements must match (3x4 matrix = 12 elements).

````python
# Convert 1D daily temperatures to a 2D weekly grid:
daily_temps = np.array([25, 26, 24, 23, 27, 28, 29, 30, 29, 28, 27, 26])
weekly_temps = daily_temps.reshape(3, 4)  # 3 weeks, 4 days each
print("2D weekly (3 weeks with 4 days each) temperature data:\n", weekly_temps)
````

````python
random_array = np.arange(10, 99, 2).reshape(5, 9)
print("Original 5x9 matrix 2Darray:\n", random_array)

print('-' * 30)

# Indexing and Slicing to work with a part of the whole data like [n:n, m:m] n=row index; m=col index
print("\nSliced - only the 3rd row:\n", random_array[2])
print("\nSliced - only 3rd column:\n", random_array[:, 2])
print("\nSliced - only the value from the 3rd row and the 3rd column:", random_array[2, 2])

print("\nSliced - from 4th row to the last:\n", random_array[3:])
print("\nSliced - from 4th column to the last:\n", random_array[:, 3:])
print("\nSliced - from 4th row & 4th column to the last:\n", random_array[3:, 3:])

print("\nSliced - upto the 3rd row:\n", random_array[:3])
print("\nSliced - upto the 3rd column:\n", random_array[:, :3])
print("\nSliced - upto the 3rd row and 3rd column:\n", random_array[:3, :3])

print("\nSliced - from 2nd to 4th row:\n", random_array[1:4])
print("\nSliced - from 2nd to 4th column:\n", random_array[:, 1:4])
print("\nSliced - from 2nd row & 2nd column to 4th row & 4th column:\n", random_array[1:4, 1:4])

print("\nSliced - the last three columns from the first two rows:\n", random_array[:2, -3:])

# Slicing extracts ranges using [start:stop:step]
print("\nSliced - every second element (for 1D array) or row (for 2D array):\n", random_array[::2])

# Boolean indexing (while a condition returns True)
print("\nExtracted - all elements > 50 into a single 1D array:", random_array[random_array > 50])
````

````python
# Get all elements from an array with conditions
temp_array = np.array(np.arange(9))
print("1D array:", temp_array)

# Filtering out the data/elements in the previous array
array_mask = temp_array >= 5
print("The elements of that array which are greater than or equal to 5:", temp_array[array_mask])
````

## Topic 2: Broadcasting
- Why Learn This?
    - Perform operations on arrays of different shapes (e.g., scaling image pixels).
- Concept:
    - NumPy "stretches" smaller arrays to match larger ones without copying data.
    - Rules: Dimensions must be compatible (either same size or one is 1).

````python
# Subtract the mean from each row of a matrix
matrix = np.array([[1, 2, 3], 
                   [4, 5, 6]])

row_means = matrix.mean(axis=1, keepdims=True)  # [[2], [5]]
print("Means of each rows:\n", row_means)
centered_matrix = matrix - row_means  # Broadcasting subtracts 2 from row 1, 5 from row 2
print("Broadcasted subtract results for each rows:\n", centered_matrix)
````

### Exercise:
Multiply a 3x1 array by a 1x3 array using broadcasting:
```python
a = np.array([[1], [2], [3]])  # Shape (3,1)
b = np.array([4, 5, 6])        # Shape (3,)
```

````python
# A 3x1 array (3 row and 1 column)
a = np.array([[1], [2], [3]])  # Shape (3,1)

# A 1x3 array (1 row and 3 columns)
b = np.array([4, 5, 6])  # Shape (1, 3)
# A 3x1 array (3 rows and 1 column) by explicitely defining "[:, np.newaxis]"
b2 = np.array([4, 5, 6])[:, np.newaxis]  # Shape (3, 1)

# Broadcasting multiplication of each row of 'a' array with the each column value of 'b' array
multiplied_array = a * b  # Makes a 3x3 multiplied matrix
# Altered behavvior in this case:
multiplied_array2 = a * b2  # Makes a 3x1 multiplied matrix

print("Multiplied 3x1 array by 1x3 array using broadcasting resultig a 3x3 matrix:\n", multiplied_array)
print("Multiplied 3x1 array by 3x1 array using broadcasting resultig a 3x1 matrix:\n", multiplied_array2)
````

# Mini-Project: Sensor Data Analysis

## Scenario:
You’re analyzing data from a wearable fitness tracker. The raw data is a 1D array of heart rates (beats per minute) recorded every minute for 24 hours (1440 values).

## Tasks:
1. Reshape the data into a 24x60 matrix (rows = hours, columns = minutes).
2. Calculate the average heart rate per hour.
3. Normalize each hour’s data by subtracting its mean.

## Sample Data:
```python
heart_rates = np.random.randint(60, 100, size=1440)  # Simulated data
```

## Hints:
- Use `reshape(24, 60)`.
- Use `axis=1` for row-wise operations.

````python
# Heart rate (BMP) data from a wearable fitness tracker for 24 Hrs. recorded every minutes
heart_rates = np.random.randint(60, 100, size=1440)  # Simulated data
heart_rates_per_hour = heart_rates.reshape(24, 60)  # Making the data into hours (row) and minutes (column) format

# Calculating the average heart rate per hour (to maintain shape for broadcasting 'keepdims=True' is used)
avg_heart_rate_per_hour = np.mean(heart_rates_per_hour, axis=1, keepdims=True)
print("Average heart rate per hour are:\n", avg_heart_rate_per_hour)

# Normalizing each hour's data by subtracting its mean
normalized_heart_rates_per_hour = heart_rates_per_hour - avg_heart_rate_per_hour
print("Normalized heart rates per hour:\n", normalized_heart_rates_per_hour)
````

#### Exercise:
- Simulate 1000 coin flips and calculate the probability of getting heads.

````python
# Simulating 1000 coin flips using 0s and 1s
coin_flips = np.random.randint(0, 2, size=1000)

# Calculating the total number of heads
# Assuming: 0 = tails, 1 = heads
# Use 'np.sum()' if the values are only 0s and 1s to get the total number of 1s as their sum, otherwise
# ...use the below method to get the count of a particular value in an NumPy array
total_heads = coin_flips[coin_flips == 1].size

# Calculating the pronanlity of getting heads
# Formula: Probablity = Total Number of Heads / Total Number of Flips
heads_probablity = (total_heads / 1000) * 100  # Multiplied by 100 to get the percentage value

print(f"Probablity of getting heads from a 1000 flips of a coin is: {heads_probablity:.1f}%")
````
