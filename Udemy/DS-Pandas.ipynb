###### `pd` Series Methods

```python
dec_num = pd.Series(data=[0.5, 1.8, 4.5, 2.7, None, 5.5, 0.8, 3.3])
```

```python
# Minimum value
dec_num.min()
```

```python
# Minimum value index
dec_num.idxmin()
```

```python
# Check for the whole data for missing value(s) and return True if a value is NaN, otherwise False
dec_num.isnull()
```

```python
# Addition of all values
dec_num.sum()
```

```python
# Returns the avarage of all the numbers
dec_num.mean()
```

### `pd` CSV file

```python
# Loads the csv file as a dataframe (use type() to check), not as a series
# Use pd.read_csv().squeeze() to make it a series for only one column as series is an one-dimensional dataframe
csv_file = pd.read_csv('TopRichestInWorld.csv', usecols=['Name']).squeeze()
csv_file
```

```python
type(csv_file)
```

```python
# To export as csv file (use index=False argument to not save an extra column for index)
csv_file.to_csv('TopRichestInWorld-Names.csv', index=False)
```
